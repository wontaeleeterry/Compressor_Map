```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# 설정
# ============================================================

# 현재 실행되는 graph.py가 있는 폴더
# BASE_DIR = Path(__file__).resolve().parent


# 주피터 노트북의 경우 아래와 같이 작성(260908)
##########################################
from pathlib import Path

# 현재 작업 중인 폴더의 절대 경로를 가져옵니다.
BASE_DIR = Path.cwd()

print(BASE_DIR)
#########################################

# 데이터 폴더
DATA_DIR = BASE_DIR / "data"


# ============================================================
# 데이터 읽기
# ============================================================

def load_data(file_path):
    """
    XDF/TXT 파일에서 데이터를 읽는다.

    파일 구조:
    1~2 : X-axis Synchronous - 1st Order
    3~4 : X-axis Overall
    5~6 : Y-axis Synchronous - 1st Order
    7~8 : Y-axis Overall
    9~10 : Z-axis Synchronous - 1st Order
    11~12 : Z-axis Overall
    13~14 : TC Speed
    """

    data = pd.read_csv(
        file_path,
        skiprows=11,
        header=None,
        usecols=range(1, 15)
    )

    data.columns = [
        "x_sord_time",
        "x_sord",

        "x_overall_time",
        "x_overall",

        "y_sord_time",
        "y_sord",

        "y_overall_time",
        "y_overall",

        "z_sord_time",
        "z_sord",

        "z_overall_time",
        "z_overall",

        "rpm_time",
        "rpm"
    ]

    # 숫자로 변환
    data = data.apply(pd.to_numeric, errors="coerce")

    # 완전히 비어 있는 행 제거
    data = data.dropna(how="all")

    return data


# ============================================================
# X-axis 그래프
# ============================================================

def plot_x_axis(data, output_file, sample_name):

    plt.figure(figsize=(12, 6))

    plt.plot(
        data["x_sord_time"],
        data["x_sord"],
        label="X-axis Synchronous - 1st Order",
        linewidth=1.2
    )

    plt.plot(
        data["x_overall_time"],
        data["x_overall"],
        label="X-axis Overall",
        linewidth=1.2
    )

    plt.xlabel("Time (sec)")
    plt.ylabel("Vibration (g)")
    plt.title(f"S/N{sample_name}: X-axis Vibration")

    plt.grid(True, alpha=0.3)
    plt.legend()

    plt.tight_layout()

    plt.savefig(
        output_file,
        dpi=150,
        bbox_inches="tight"
    )

    plt.close()


# ============================================================
# Y-axis 그래프
# ============================================================

def plot_y_axis(data, output_file, sample_name):

    plt.figure(figsize=(12, 6))

    plt.plot(
        data["y_sord_time"],
        data["y_sord"],
        label="Y-axis Synchronous - 1st Order",
        linewidth=1.2
    )

    plt.plot(
        data["y_overall_time"],
        data["y_overall"],
        label="Y-axis Overall",
        linewidth=1.2
    )

    plt.xlabel("Time (sec)")
    plt.ylabel("Vibration (g)")
    plt.title(f"S/N{sample_name}: Y-axis Vibration")

    plt.grid(True, alpha=0.3)
    plt.legend()

    plt.tight_layout()

    plt.savefig(
        output_file,
        dpi=150,
        bbox_inches="tight"
    )

    plt.close()


# ============================================================
# Z-axis 그래프
# ============================================================

def plot_z_axis(data, output_file, sample_name):

    plt.figure(figsize=(12, 6))

    plt.plot(
        data["z_sord_time"],
        data["z_sord"],
        label="Z-axis Synchronous - 1st Order",
        linewidth=1.2
    )

    plt.plot(
        data["z_overall_time"],
        data["z_overall"],
        label="Z-axis Overall",
        linewidth=1.2
    )

    plt.xlabel("Time (sec)")
    plt.ylabel("Vibration (g)")
    plt.title(f"S/N{sample_name}: Z-axis Vibration")

    plt.grid(True, alpha=0.3)
    plt.legend()

    plt.tight_layout()

    plt.savefig(
        output_file,
        dpi=150,
        bbox_inches="tight"
    )

    plt.close()


# ============================================================
# TC Speed 그래프
# ============================================================

def plot_rpm(data, output_file, sample_name):

    plt.figure(figsize=(12, 6))

    plt.plot(
        data["rpm_time"],
        data["rpm"],
        label="TC Speed",
        linewidth=1.5
    )

    plt.xlabel("Time (sec)")
    plt.ylabel("Speed (rpm)")
    plt.title(f"S/N{sample_name}: TC Speed")

    plt.grid(True, alpha=0.3)
    plt.legend()

    plt.tight_layout()

    plt.savefig(
        output_file,
        dpi=150,
        bbox_inches="tight"
    )

    plt.close()


# ============================================================
# 개별 파일 처리
# ============================================================

def process_file(file_path):

    print(f"Processing: {file_path}")

    try:

        # 데이터 읽기
        data = load_data(file_path)

        # 파일명
        sample_name = file_path.stem

        # 파일이 존재하는 폴더
        output_dir = file_path.parent

        # --------------------------------------------
        # 1. X-axis
        # --------------------------------------------

        x_output = (
            output_dir /
            f"{sample_name}_X-axis_Vibration.png"
        )

        plot_x_axis(
            data,
            x_output,
            sample_name
        )

        # --------------------------------------------
        # 2. Y-axis
        # --------------------------------------------

        y_output = (
            output_dir /
            f"{sample_name}_Y-axis_Vibration.png"
        )

        plot_y_axis(
            data,
            y_output,
            sample_name
        )

        # --------------------------------------------
        # 3. Z-axis
        # --------------------------------------------

        z_output = (
            output_dir /
            f"{sample_name}_Z-axis_Vibration.png"
        )

        plot_z_axis(
            data,
            z_output,
            sample_name
        )

        # --------------------------------------------
        # 4. TC Speed
        # --------------------------------------------

        rpm_output = (
            output_dir /
            f"{sample_name}_TC_Speed.png"
        )

        plot_rpm(
            data,
            rpm_output,
            sample_name
        )

        print("  -> X-axis graph created")
        print("  -> Y-axis graph created")
        print("  -> Z-axis graph created")
        print("  -> TC Speed graph created")
        print()

    except Exception as e:

        print(f"  ERROR: {e}")
        print()


# ============================================================
# 전체 폴더 검색
# ============================================================

def main():

    print("=" * 60)
    print("Vibration Graph Generator")
    print("=" * 60)

    print(f"Data directory: {DATA_DIR}")
    print()

    if not DATA_DIR.exists():

        print(
            f"ERROR: Data directory does not exist:\n"
            f"{DATA_DIR}"
        )

        return

    # --------------------------------------------------------
    # ./data 아래의 모든 TXT 파일 검색
    # --------------------------------------------------------

    txt_files = sorted(
        DATA_DIR.glob("*/*.txt")
    )

    if not txt_files:

        print("No TXT files found.")

        return

    print(
        f"Found {len(txt_files)} TXT file(s).\n"
    )

    # --------------------------------------------------------
    # 각각의 파일 처리
    # --------------------------------------------------------

    for file_path in txt_files:

        process_file(file_path)

    print("=" * 60)
    print("All files processed.")
    print("=" * 60)


# ============================================================
# 프로그램 시작
# ============================================================

if __name__ == "__main__":
    main()
```
